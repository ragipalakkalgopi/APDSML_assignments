```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = sns.load_dataset('titanic')
df.head()
```

```python
sns.countplot(data=df, x='embarked', hue='sex')
plt.title('Passenger Count by Embarked Port and Gender')
plt.xlabel('Embarked Port')
plt.ylabel('Count')
plt.show()
```

```python
sns.swarmplot(data=df, x='pclass', y='age', hue='sex')
plt.title('Age Distribution across Passenger Classes')
plt.xlabel('Passenger Class')
plt.ylabel('Age')
plt.show()
```

```python
gender_counts = df['sex'].value_counts()

plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%', explode=(0.1, 0))
plt.title('Gender Distribution of Passengers')
plt.show()
```

```python
corr = df[['age', 'fare', 'pclass', 'sibsp', 'parch', 'survived']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

```python
sns.violinplot(data=df, x='pclass', y='fare', hue='sex')
plt.title('Fare Distribution by Class and Gender')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.show()
```

```python
sns.scatterplot(data=df, x='age', y='fare', hue='survived')
plt.title('Age vs Fare (Survival)')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```
